```python
# Dong goi ket qua -> /kaggle/working/results_deep_kt.zip (tai ve o tab Output)
import zipfile, pathlib
repo = pathlib.Path("/kaggle/working/repo")
files = [p for pat in ['reports/tables/b1_deep_*', 'lmsrag/artifacts/v2/predictions/b1_deep_*'] for p in repo.glob(pat)]
with zipfile.ZipFile("/kaggle/working/results_deep_kt.zip", "w", zipfile.ZIP_DEFLATED) as z:
    for p in files:
        z.write(p, p.relative_to(repo).as_posix())
print("\n".join(sorted(p.relative_to(repo).as_posix() for p in files)))
```
