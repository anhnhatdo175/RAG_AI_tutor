```python
# Dong goi ket qua -> /kaggle/working/results_lightgcn.zip (tai ve o tab Output)
import zipfile, pathlib
repo = pathlib.Path("/kaggle/working/repo")
files = [p for pat in ['reports/tables/b4_lightgcn_*', 'lmsrag/artifacts/v2/predictions/b4_lightgcn_*'] for p in repo.glob(pat)]
with zipfile.ZipFile("/kaggle/working/results_lightgcn.zip", "w", zipfile.ZIP_DEFLATED) as z:
    for p in files:
        z.write(p, p.relative_to(repo).as_posix())
print("\n".join(sorted(p.relative_to(repo).as_posix() for p in files)))
```
